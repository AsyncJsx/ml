```
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
```

```
df = pd.read_csv('p3_synthetic.csv')
X = df.drop(columns='target')
y = df.target
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.3, random_state=6)
models = {
    'No control': DecisionTreeClassifier(random_state=0),
    'max_depth=2': DecisionTreeClassifier(max_depth=2, random_state=0),
    'ccp_alpha=0.15': DecisionTreeClassifier(ccp_alpha=0.15, random_state=0),
}
for n, m in models.items():
    m.fit(Xtr, ytr)
    print('%-15s train=%.3f test=%.3f depth=%d leaves=%d' % (n, m.score(Xtr, ytr), m.score(Xte, yte), m.get_depth(), m.get_n_leaves()))
```

```
plt.figure(figsize=(16, 7))
plot_tree(models['max_depth=2'], feature_names=X.columns, class_names=['0', '1'], filled=True)
plt.show()
```

```
depths = range(1, 11)
tr, te = [], []
for d in depths:
    m = DecisionTreeClassifier(max_depth=d, random_state=0).fit(Xtr, ytr)
    tr.append(m.score(Xtr, ytr))
    te.append(m.score(Xte, yte))
plt.plot(depths, tr, label='train')
plt.plot(depths, te, label='test')
plt.xlabel('max_depth')
plt.ylabel('accuracy')
plt.title('Overfitting')
plt.legend()
plt.show()
```
