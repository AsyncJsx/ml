```
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_curve, roc_auc_score, accuracy_score
```

```
df = pd.read_csv('p7_cancer.csv')
X = df.drop(columns='target')
y = df.target
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
z = np.linspace(-8, 8, 200)
plt.plot(z, 1 / (1 + np.exp(-z)))
plt.axhline(0.5, color='r', ls='--')
plt.title('Sigmoid')
plt.show()
```

```
c = ['mean_radius', 'mean_texture']
m2 = make_pipeline(StandardScaler(), LogisticRegression()).fit(Xtr[c], ytr)
gx, gy = np.meshgrid(np.linspace(X[c[0]].min(), X[c[0]].max(), 200), np.linspace(X[c[1]].min(), X[c[1]].max(), 200))
Z = m2.predict(pd.DataFrame(np.c_[gx.ravel(), gy.ravel()], columns=c)).reshape(gx.shape)
plt.contourf(gx, gy, Z, alpha=.3)
plt.scatter(X[c[0]], X[c[1]], c=y, edgecolor='k', s=15)
plt.xlabel(c[0])
plt.ylabel(c[1])
plt.title('Decision boundary')
plt.show()
```

```
m = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000)).fit(Xtr, ytr)
pr = m.predict_proba(Xte)[:, 1]
fpr, tpr, _ = roc_curve(yte, pr)
auc = roc_auc_score(yte, pr)
print('accuracy = %.3f  AUC = %.3f' % (accuracy_score(yte, m.predict(Xte)), auc))
plt.plot(fpr, tpr, label='AUC=%.3f' % auc)
plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel('FPR')
plt.ylabel('TPR')
plt.legend()
plt.show()
```
