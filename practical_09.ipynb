```
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier
from xgboost import XGBClassifier
```

```
df = pd.read_csv('p9_cancer.csv')
X = df.drop(columns='target')
y = df.target
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
models = {
    'Random Forest (Bagging)': RandomForestClassifier(200, random_state=42),
    'AdaBoost': AdaBoostClassifier(n_estimators=200, random_state=42),
    'XGBoost': XGBClassifier(n_estimators=200, eval_metric='logloss', random_state=42),
}
fig, ax = plt.subplots(1, 3, figsize=(20, 5))
for a, (n, m) in zip(ax, models.items()):
    cv = cross_val_score(m, Xtr, ytr, cv=3).mean()
    m.fit(Xtr, ytr)
    print('%-24s cv=%.3f test=%.3f' % (n, cv, m.score(Xte, yte)))
    s = pd.Series(m.feature_importances_, index=X.columns).nlargest(10)
    s[::-1].plot.barh(ax=a, title=n)
plt.tight_layout()
plt.show()
```
