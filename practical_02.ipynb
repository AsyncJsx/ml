```
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```
df = pd.read_csv('p2_housing.csv')
y = df.price.values

def ols(X, y):
    A = np.c_[np.ones(len(X)), X]
    return np.linalg.lstsq(A, y, rcond=None)[0]

def pred(X, w):
    return np.c_[np.ones(len(X)), X] @ w

def rmse(y, p):
    return np.sqrt(np.mean((y - p) ** 2))

def r2(y, p):
    return 1 - ((y - p) ** 2).sum() / ((y - y.mean()) ** 2).sum()

sets = {'Simple': df[['area']].values, 'Multiple': df[['area', 'bedrooms', 'age']].values}
fitted = {}
for name, X in sets.items():
    p = pred(X, ols(X, y))
    q = LinearRegression().fit(X, y).predict(X)
    fitted[name] = p
    print(name, 'NumPy   RMSE=%.2f R2=%.4f' % (rmse(y, p), r2(y, p)))
    print(name, 'sklearn RMSE=%.2f R2=%.4f' % (np.sqrt(mean_squared_error(y, q)), r2_score(y, q)))
```

```
fig, ax = plt.subplots(1, 3, figsize=(16, 4))
o = np.argsort(sets['Simple'][:, 0])
ax[0].scatter(df.area, y, s=12)
ax[0].plot(df.area.values[o], fitted['Simple'][o], 'r')
ax[0].set_title('Simple regression line')
for a, name in zip(ax[1:], ['Simple', 'Multiple']):
    a.scatter(fitted[name], y - fitted[name], s=12)
    a.axhline(0, color='r')
    a.set_title(name + ' residuals')
    a.set_xlabel('fitted')
plt.show()
```
